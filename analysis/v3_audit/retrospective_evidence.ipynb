# Retrospective saved-prediction evidence
This notebook executes the NumPy-only recomputation script; it does not retrain models. All periods were previously exposed. Intervals are conditional retrospective postselection uncertainty, not selection-adjusted confidence or a private-leaderboard forecast.

Each displayed scope normalizes blend components over that complete scope. The month-66 exclusion retains the original late-block prediction scales and calendar positions during block resampling. Full definitions, input hashes, checks and limitations are in `artifacts/v3_audit/retrospective_metrics.json`.


```python
from pathlib import Path
import json
import sys
project = Path.cwd().resolve()
while not (project / 'analysis/v3_audit/recompute_evidence.py').exists():
    if project == project.parent: raise FileNotFoundError('Run inside this project')
    project = project.parent
sys.path.insert(0, str(project / 'analysis/v3_audit'))
from recompute_evidence import run
evidence = run()
print(evidence['status'])
print(evidence['uncertainty_label'])
```

```python
for scope, result in evidence['scopes'].items():
    print(scope, 'rows=', result['rows'], 'months=', result['scored_months'])
    for model, score in result['cosine'].items():
        print(f'  {model:26s} {score:.9f}')
```

```python
for row in evidence['paired_comparisons']:
    if row['candidate'] == 'blend_q1p1' and row['baseline'] == 'path_capacity_q1p2':
        print(row['scope'], 'block=', row['block_months'],
              'delta=', round(row['point_delta_cosine'], 6),
              '95% interval=', [round(row[k], 6) for k in ['percentile_95_lower', 'percentile_95_upper']])
print('Exact alignment and finite checks:', evidence['checks'])
```
